## Combining async concepts with rate limiting.

*Simple Exercise :*

 Implement an async fetcher that:   
 - Reads a list of URLs.   
 - Fetches them concurrently with a max concurrency of 10.   
 - Retries 5xx errors with exponential backoff (max 3 retries).   
 - Enforces a global rate limit of 50 requests/minute.   
 - Returns JSON responses and logs per-URL latency and retry count.
